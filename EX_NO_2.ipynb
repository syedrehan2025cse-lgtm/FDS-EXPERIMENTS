```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the dataset
file_path = r"C:\sales-data.csv"
df = pd.read_csv(file_path)

# 2. Display the first five rows
print("First five rows:")
print(df.head())

# 3. Check missing values
print("\nMissing values:")
print(df.isnull().sum())

# 4. Data preprocessing
df['Sales'] = df['Sales'].fillna(df['Sales'].mean())

df.dropna(subset=['Product', 'Quantity', 'Region', 'Date'],
          inplace=True)

df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df.dropna(subset=['Date'], inplace=True)

# 5. Display summary statistics
print("\nSummary statistics:")
print(df.describe())

# 6. Group by Product and calculate total Sales
# and Quantity
product_summary = df.groupby('Product').agg({
    'Sales': 'sum',
    'Quantity': 'sum'
}).reset_index()

print("\nProduct summary:")
print(product_summary)

# 7. Bar plot: Total sales by product
plt.figure(figsize=(10, 6))
plt.bar(product_summary['Product'],
        product_summary['Sales'])
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.title('Total Sales by Product')
plt.tight_layout()
plt.show()

# 8. Line plot: Sales over time
sales_over_time = df.groupby('Date')['Sales'].sum().reset_index()

plt.figure(figsize=(10, 6))
plt.plot(sales_over_time['Date'],
         sales_over_time['Sales'], marker='o')
plt.xlabel('Date')
plt.ylabel('Total Sales')
plt.title('Sales Over Time')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 9. Pivot table: Sales by Region and Product
pivot_table = df.pivot_table(
    values='Sales',
    index='Region',
    columns='Product',
    aggfunc='sum',
    fill_value=0
)

print("\nSales by Region and Product:")
print(pivot_table)

# 10. Correlation matrix (numeric columns only)
correlation_matrix = df[['Sales', 'Quantity']].corr()

print("\nCorrelation matrix:")
print(correlation_matrix)

# 11. Heatmap of the correlation matrix
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix,
            annot=True, cmap='coolwarm',
            vmin=-1, vmax=1)
plt.title('Correlation Matrix')
plt.tight_layout()
plt.show()
```
