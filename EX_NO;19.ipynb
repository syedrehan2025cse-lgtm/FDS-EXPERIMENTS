```python
import numpy as np
import pandas as pd

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix, classification_report

# Load the Iris dataset
iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=[
        'sepal.length',
        'sepal.width',
        'petal.length',
        'petal.width'
    ]
)

df['variety'] = pd.Categorical.from_codes(
    iris.target,
    ['Setosa', 'Versicolor', 'Virginica']
)

# Display dataset information
df.info()

# Display the number of flowers in each variety
print("\nVariety Counts:")
print(df['variety'].value_counts())

# Display first five rows
print("\nFirst Five Rows:")
print(df.head())

# Separate features and labels
features = df.iloc[:, :-1].values
label = df.iloc[:, 4].values

# Split dataset into training and testing sets
xtrain, xtest, ytrain, ytest = train_test_split(
    features,
    label,
    test_size=0.2,
    random_state=42
)

# Create and train KNN model
model_KNN = KNeighborsClassifier(n_neighbors=5)
model_KNN.fit(xtrain, ytrain)

# Display model information
print("\nKNN Model:")
print(model_KNN)

# Display training and testing accuracy
print("\nTraining Accuracy:")
print(model_KNN.score(xtrain, ytrain))

print("\nTesting Accuracy:")
print(model_KNN.score(xtest, ytest))

# Predict the full dataset
ypred = model_KNN.predict(features)

# Display confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(
    label,
    ypred,
    labels=['Setosa', 'Versicolor', 'Virginica']
))

# Display classification report
print("\nClassification Report:")
print(classification_report(
    label,
    ypred,
    labels=['Setosa', 'Versicolor', 'Virginica'],
    zero_division=0
))

```
