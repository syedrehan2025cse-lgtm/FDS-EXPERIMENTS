```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the dataset
diabetes_df = pd.read_csv("diabetes.csv")

# 2. Display the first five rows
print("First five rows:")
print(diabetes_df.head())

# 3. Display the last five rows
print("\nLast five rows:")
print(diabetes_df.tail())

# 4. Display dataset shape
print("\nDataset shape:")
print(diabetes_df.shape)

# 5. Display column names
print("\nColumn names:")
print(diabetes_df.columns)

# 6. Display dataset information
print("\nDataset information:")
diabetes_df.info()

# 7. Display summary statistics
print("\nSummary statistics:")
print(diabetes_df.describe())

# 8. Check missing values
print("\nMissing values:")
print(diabetes_df.isnull().sum())

# 9. Check duplicate rows
print("\nDuplicate rows:")
print(diabetes_df.duplicated().sum())

# 10. Display correlation between numerical columns
print("\nCorrelation matrix:")
print(diabetes_df.corr(numeric_only=True))

# 11. Visualize the Outcome column
if 'Outcome' in diabetes_df.columns:
    sns.countplot(data=diabetes_df, x='Outcome')
    plt.title("Diabetes Outcome Distribution")
    plt.xlabel("Outcome (0 = No Diabetes, 1 = Diabetes)")
    plt.ylabel("Count")
    plt.show()

# 12. Display a heatmap of correlations
plt.figure(figsize=(10, 6))
sns.heatmap(
    diabetes_df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm"
)
plt.title("Correlation Heatmap of Diabetes Dataset")
plt.tight_layout()
plt.show()
```
