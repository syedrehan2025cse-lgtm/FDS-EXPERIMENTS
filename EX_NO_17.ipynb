```python


# Exp No: 17
# Experiment to understand Linear Regression for a given dataset

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import pickle

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# Create the Salary dataset
data = {
    'YearsExperience': [
        1.1, 1.3, 1.5, 2.0, 2.2, 2.9, 3.0, 3.2, 3.2, 3.7,
        3.9, 4.0, 4.0, 4.1, 4.5, 4.9, 5.1, 5.3, 5.9, 6.0,
        6.8, 7.1, 7.9, 8.2, 8.7, 9.0, 9.5, 9.6, 10.3, 10.5
    ],
    'Salary': [
        39343, 46205, 37731, 43525, 39891, 56642, 60150, 54445, 64445, 57189,
        63218, 55794, 56957, 57081, 61111, 67938, 66029, 83088, 81363, 93940,
        91738, 98273, 101302, 113812, 109431, 105582, 116969, 112635, 122391, 121872
    ]
}

df = pd.DataFrame(data)

# Display the dataset
print(df)

# Display dataset information
df.info()

# Remove missing values
df.dropna(inplace=True)
df.info()

# Display statistical summary
print(df.describe())

# Separate features and label
features = df.iloc[:, [0]].values
label = df.iloc[:, [1]].values

# Split the dataset into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(
    features, label, test_size=0.2, random_state=42
)

# Create and train the Linear Regression model
model = LinearRegression()
model.fit(x_train, y_train)

# Display the model
print(model)

# Calculate training and testing accuracy (R-squared score)
print("Training Score:", model.score(x_train, y_train))
print("Testing Score:", model.score(x_test, y_test))

# Predict salary for the test data
y_pred = model.predict(x_test)
print("Predicted Salaries:")
print(y_pred)

# Display model coefficients
print("Coefficient:", model.coef_)
print("Intercept:", model.intercept_)

# Plot the dataset and regression line
plt.scatter(features, label, color='blue', label='Actual Salary')
plt.plot(features, model.predict(features), color='red', label='Regression Line')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.title('Linear Regression - Salary Prediction')
plt.legend()
plt.show()

# Save the trained model
with open('linear_regression_model.pkl', 'wb') as file:
    pickle.dump(model, file)

# Load the saved model
with open('linear_regression_model.pkl', 'rb') as file:
    loaded_model = pickle.load(file)

# Predict salary for user input
years = float(input("Enter Years of Experience: "))
estimated_salary = loaded_model.predict(np.array([[years]]))

print("Estimated Salary for", years, "years of experience:", estimated_salary[0][0])

```
