```python
import numpy as np
import pandas as pd

# 1. Create and display the dataset
data = {
    'Country': ['France', 'Spain', 'Germany', 'Spain', 'Germany',
                'France', 'Spain', 'France', np.nan, 'France'],
    'Age': [44.0, 27.0, 30.0, 38.0, 40.0, 35.0, np.nan, 48.0, 50.0, 37.0],
    'Salary': [72000.0, 48000.0, 54000.0, 61000.0, np.nan,
               58000.0, 52000.0, 79000.0, 83000.0, 67000.0],
    'Purchased': ['No', 'Yes', 'No', 'No', 'Yes',
                  'Yes', 'No', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)

print("OUTPUT 1: ORIGINAL DATASET")
display(df)

# 2. Display dataset information
print("OUTPUT 2: DATASET INFORMATION")
df.info()

# 3. Display Country column
print("\nOUTPUT 3: COUNTRY COLUMN")
display(df[['Country']])

# 4. Find the mode of Country
print("OUTPUT 4: MODE OF COUNTRY")
display(df.Country.mode())

# 5. Display the first mode value
print("OUTPUT 5: FIRST MODE VALUE")
print(df.Country.mode()[0])

# 6. Display the type of mode
print("OUTPUT 6: TYPE OF MODE")
print(type(df.Country.mode()))

# 7. Handle missing values and display the cleaned dataset
df['Country'] = df['Country'].fillna(df['Country'].mode()[0])
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Salary'] = df['Salary'].fillna(round(df['Salary'].mean()))

print("OUTPUT 7: DATASET AFTER HANDLING MISSING VALUES")
display(df)

# 8. Display one-hot encoded Country columns
print("OUTPUT 8: ONE-HOT ENCODING")
display(pd.get_dummies(df.Country))

# Create the updated dataset
updated_dataset = pd.concat(
    [pd.get_dummies(df.Country), df.iloc[:, [1, 2, 3]]],
    axis=1
)

print("UPDATED DATASET AFTER COUNTRY ENCODING")
display(updated_dataset)

# Display the cleaned dataframe information
print("DATAFRAME INFORMATION AFTER CLEANING")
df.info()

# Convert Purchased: No = 0 and Yes = 1
updated_dataset['Purchased'] = updated_dataset['Purchased'].replace(
    {'No': 0, 'Yes': 1}
)

print("FINAL UPDATED DATASET")
display(updated_dataset)

print("EXPERIMENT NO. 15 COMPLETED SUCCESSFULLY")

```
