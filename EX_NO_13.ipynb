```python
import numpy as np
import pandas as pd

# Experiment No. 13
# Handling Missing and Inappropriate Data in a Dataset

# 1. Create Hotel Dataset
data = {
    'CustomerID': [1, 2, 3, 4, 5, 6, 7, 8, 9, 9, 10],
    'Age_Group': [
        '20-25', '30-35', '25-30', '20-25', '35+',
        '35+', '35+', '20-25', '25-30', '25-30', '30-35'
    ],
    'Rating(1-5)': [4, 5, 6, -1, 3, 3, 4, 7, 2, 2, 5],
    'Hotel': [
        'Ibis', 'LemonTree', 'RedFox', 'LemonTree', 'Ibis',
        'Ibys', 'RedFox', 'LemonTree', 'Ibis', 'Ibis', 'RedFox'
    ],
    'FoodPreference': [
        'veg', 'Non-Veg', 'Veg', 'Veg', 'Vegetarian',
        'Non-Veg', 'Vegetarian', 'Veg', 'Non-Veg', 'Non-Veg', 'non-Veg'
    ],
    'Bill': [1300, 2000, 1322, 1234, 989, 1909, 1000, 2999, 3456, 3456, 6755],
    'NoOfPax': [2, 3, 2, 2, 2, 2, -1, -10, 3, 3, 4],
    'EstimatedSalary': [
        40000, 59000, 30000, 120000, 45000,
        122220, 21122, 345673, -99999, -99999, 87777
    ],
    'Age_Group.1': [
        '20-25', '30-35', '25-30', '20-25', '35+',
        '35+', '35+', '20-25', '25-30', '25-30', '30-35'
    ]
}

df = pd.DataFrame(data)

print("ORIGINAL HOTEL DATASET")
display(df)

# 2. Identify duplicate rows
print("DUPLICATE ROWS")
display(df.duplicated())

# 3. Display DataFrame information
print("DATAFRAME INFORMATION")
df.info()

# 4. Remove duplicate rows
df.drop_duplicates(inplace=True)

print("DATASET AFTER REMOVING DUPLICATES")
display(df)

print("Number of rows:", len(df))

# 5. Reset index
index = np.array(list(range(0, len(df))))
df.set_index(index, inplace=True)

print("RESET INDEX")
display(df)

# 6. Remove unnecessary column
df.drop(['Age_Group.1'], axis=1, inplace=True)

print("AFTER DROPPING EXTRA COLUMN")
display(df)

# 7. Handle inappropriate numerical values
df.loc[df['CustomerID'] < 0, 'CustomerID'] = np.nan
df.loc[df['Bill'] < 0, 'Bill'] = np.nan
df.loc[df['EstimatedSalary'] < 0, 'EstimatedSalary'] = np.nan

df.loc[
    (df['NoOfPax'] < 1) | (df['NoOfPax'] > 20),
    'NoOfPax'
] = np.nan

# Ratings must be between 1 and 5
df.loc[
    (df['Rating(1-5)'] < 1) | (df['Rating(1-5)'] > 5),
    'Rating(1-5)'
] = np.nan

print("AFTER REPLACING INAPPROPRIATE VALUES WITH NaN")
display(df)

# 8. Display unique values
print("UNIQUE AGE GROUPS")
print(df['Age_Group'].unique())

print("UNIQUE HOTELS")
print(df['Hotel'].unique())

# 9. Correct inappropriate hotel name
df['Hotel'] = df['Hotel'].replace({'Ibys': 'Ibis'})

# 10. Standardize food preference values
df['FoodPreference'] = df['FoodPreference'].replace({
    'Vegetarian': 'Veg',
    'veg': 'Veg',
    'non-Veg': 'Non-Veg'
})

print("AFTER CORRECTING CATEGORICAL VALUES")
display(df)

# 11. Fill missing numerical values
df['EstimatedSalary'] = df['EstimatedSalary'].fillna(
    round(df['EstimatedSalary'].mean())
)

df['NoOfPax'] = df['NoOfPax'].fillna(
    round(df['NoOfPax'].median())
)

df['Rating(1-5)'] = df['Rating(1-5)'].fillna(
    round(df['Rating(1-5)'].median())
)

df['Bill'] = df['Bill'].fillna(
    round(df['Bill'].mean())
)

# 12. Fill missing categorical values with mode
for column in ['Age_Group', 'Hotel', 'FoodPreference']:
    df[column] = df[column].fillna(df[column].mode()[0])

print("FINAL CLEANED DATASET")
display(df)

# 13. Check for missing values
print("MISSING VALUES IN EACH COLUMN")
print(df.isnull().sum())

print("EXPERIMENT NO. 13 COMPLETED SUCCESSFULLY")

```
