```python
import pandas as pd
import matplotlib.pyplot as plt
data = {
    "Year": [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024],
    "Job_Postings": [5000, 6200, 7800, 9800, 12500, 14000, 18000, 22000, 25000, 29000]
}

df = pd.DataFrame(data)
print(df)
df.plot(x="Year", y="Job_Postings", marker="o")
plt.title("Data Science Job Posting Trend")
plt.show()



import pandas as pd
import matplotlib.pyplot as plt
data = {
    "Role": ["Data Scientist", "Data Analyst", "Data Engineer", "ML Engineer"],
    "Count": [120, 90, 70, 50]
}

df = pd.DataFrame(data)
print(df)
df.plot(x="Role", y="Count", kind="bar", legend=False)
plt.title("Distribution of Data Science Roles")
plt.ylabel("Count")
plt.show()
df.set_index("Role").plot(kind="pie", y="Count", autopct="%1.1f%%")
plt.title("Role Distribution")
plt.ylabel("")
plt.show()




import pandas as pd
data = {
    "ID": [1, 2, 3],
    "Name": ["John", "Alice", "Bob"],
    "Marks": [85, 90, 88]
}
df = pd.DataFrame(data)
print("Structured Data")
print(df)
semi = {
    "ID": 1,
    "Name": "John",
    "Marks": 85
}
print("\nSemi-Structured Data")
print(semi)
text = "Data Science is growing rapidly."
print("\nUnstructured Data")
print(text)



from cryptography.fernet import Fernet
key = Fernet.generate_key()
f = Fernet(key)
data = "Hello World"
encrypted = f.encrypt(data.encode())
print("Encrypted:", encrypted)
decrypted = f.decrypt(encrypted).decode()
print("Decrypted:", decrypted)
```
