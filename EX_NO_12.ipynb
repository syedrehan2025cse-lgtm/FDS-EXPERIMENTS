```python

import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Experiment No. 12
# Experiment to Detect Outliers in a Given Dataset

# Generate 16 random numbers between 1 and 100
array = np.random.randint(1, 100, 16)

print("Original Array:")
print(array)

# Calculate mean
print("\nMean:")
print(array.mean())

# Calculate percentiles
print("\n25th Percentile (Q1):")
print(np.percentile(array, 25))

print("\n50th Percentile (Median):")
print(np.percentile(array, 50))

print("\n75th Percentile (Q3):")
print(np.percentile(array, 75))

print("\n100th Percentile (Maximum):")
print(np.percentile(array, 100))

# Function to detect outlier limits using IQR
def outDetection(array):
    Q1, Q3 = np.percentile(array, [25, 75])
    IQR = Q3 - Q1
    lr = Q1 - (1.5 * IQR)
    ur = Q3 + (1.5 * IQR)
    return lr, ur

# Calculate lower and upper ranges
lr, ur = outDetection(array)

print("\nLower Range:", lr)
print("Upper Range:", ur)

# Plot original data distribution
sns.displot(array)
plt.title("Distribution of Original Data")
plt.show()

# Plot histogram with density curve
sns.histplot(array, kde=True)
plt.title("Original Data Distribution")
plt.show()

# Remove outliers
new_array = array[(array >= lr) & (array <= ur)]

print("\nArray After Removing Outliers:")
print(new_array)

# Plot data after removing outliers
sns.displot(new_array)
plt.title("Distribution After Removing Outliers")
plt.show()

# Calculate new lower and upper ranges
lr1, ur1 = outDetection(new_array)

print("\nNew Lower Range:", lr1)
print("New Upper Range:", ur1)

# Final array without outliers
final_array = new_array[
    (new_array >= lr1) & (new_array <= ur1)
]

print("\nFinal Array:")
print(final_array)

# Plot final distribution
sns.histplot(final_array, kde=True)
plt.title("Final Data Distribution")
plt.show()

print("\nExperiment No. 12 Completed Successfully!")
```
