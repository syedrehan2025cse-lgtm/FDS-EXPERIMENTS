```python
import numpy as np
import pandas as pd

# Experiment No. 11: Pandas DataFrame

# Create DataFrame
list1 = [[1, 'Smith', 50000],
         [2, 'Jones', 60000]]

df1 = pd.DataFrame(list1)
display(df1)

df1.columns = ['Empd', 'Name', 'Salary']
display(df1)
df1.info()

# Employee dataset
df = pd.DataFrame({
    'emp id': [1, 2, 3, 4, 5, 6, 7],
    'name ': [
        'SREE VARSSINI K S',
        'SREEMATHI B',
        'SREYA G',
        'SREYASKARI MULLAPUDI',
        'SRI AKASH U G',
        'SRI HARSHAVARDHANAN R',
        'SRI HARSHAVARDHANAN R'
    ],
    'salary': [5000, 6000, 7000, 5000, 8000, 3000, 6000]
})

# Display first and last five rows
display(df.head())
display(df.tail())

# DataFrame information
df.info()

# Salary column
display(df.salary)
print(type(df.salary))

# Statistical calculations
print("Mean:", df.salary.mean())
print("Median:", df.salary.median())
print("Mode:")
display(df.salary.mode())

print("Variance:", df.salary.var())
print("Standard Deviation:", df.salary.std())

# Statistical descriptions
display(df.describe())
display(df.describe(include='all'))

# Columns
empCol = df.columns
print(empCol)

# Convert DataFrame to NumPy array
emparray = df.values
print(emparray)

# Convert array back to DataFrame
employee_DF = pd.DataFrame(emparray, columns=empCol)
display(employee_DF)

print("Experiment No. 11 completed successfully!")
```
