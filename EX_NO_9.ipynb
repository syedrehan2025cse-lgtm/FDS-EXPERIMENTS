```python
# Experiment to understand Matplotlib library use cases
# in Data Science through visualization using Iris Dataset

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

%matplotlib inline

# Load Iris dataset
data = sns.load_dataset('iris')

# Change column names to match your experiment
data.columns = [
    'SepalLengthCm',
    'SepalWidthCm',
    'PetalLengthCm',
    'PetalWidthCm',
    'variety'
]

# Add Id column
data.insert(0, 'Id', range(1, 151))

# Display dataset
data

# Display information
data.info()

# Statistical description
data.describe()

# Count of each variety
data.value_counts('variety')

# Count plot
sns.countplot(x='variety', data=data)
plt.show()

# Convert categorical data into dummy variables
dummies = pd.get_dummies(data.variety)

FinalDataset = pd.concat(
    [pd.get_dummies(data.variety),
     data.iloc[:, [0, 1, 2, 3]]],
    axis=1
)

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Load Iris dataset
data = sns.load_dataset('iris')

# Rename columns
data.columns = [
    'SepalLengthCm', 'SepalWidthCm',
    'PetalLengthCm', 'PetalWidthCm', 'variety'
]

# Add ID column
data.insert(0, 'Id', range(1, 151))

# Display dataset
display(data)

# Dataset information
data.info()

# Statistical summary
display(data.describe())

# Count of each species
print(data.value_counts('variety'))

# Count plot
sns.countplot(x='variety', data=data)
plt.show()

# Dummy variables
dummies = pd.get_dummies(data['variety'])

FinalDataset = pd.concat(
    [dummies, data.iloc[:, [0, 1, 2, 3]]],
    axis=1
)
display(FinalDataset.head())

# Sepal scatter plot
sns.scatterplot(
    x='SepalLengthCm',
    y='SepalWidthCm',
    hue='variety',
    data=data
)
plt.show()

# Petal scatter plot
sns.scatterplot(
    x='PetalLengthCm',
    y='PetalWidthCm',
    hue='variety',
    data=data
)
plt.show()

# Pair plot
sns.pairplot(data, hue='variety', height=3)
plt.show()

# Histograms for all four features
for feature in [
    'PetalLengthCm',
    'PetalWidthCm',
    'SepalLengthCm',
    'SepalWidthCm'
]:
    sns.FacetGrid(
        data, hue='variety', height=5
    ).map(
        sns.histplot, feature
    ).add_legend()
    plt.show()
```
