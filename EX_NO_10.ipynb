```python
import numpy as np

# Create 1-D array using values from the dataset
array = np.array([15, 15, 16, 16, 17, 17, 18, 18, 19])

print("1. Original Array:")
print(array)

# Square root
print("\n2. Square Root of Array:")
print(np.sqrt(array))

# Number of dimensions
print("\n3. Number of Dimensions:")
print(array.ndim)

# Reshape into 3x3
new_array = array.reshape(3, 3)

print("\n4. Reshaped Array:")
print(new_array)

# Number of dimensions after reshape
print("\n5. Number of Dimensions after Reshape:")
print(new_array.ndim)

# Ravel
print("\n6. Ravelled Array:")
print(new_array.ravel())

# Reshape again
newm = new_array.reshape(3, 3)

print("\n7. New Array:")
print(newm)

# Slicing operations
print("\n8. newm[2,1:3]:")
print(newm[2,1:3])

print("\n9. newm[1:2,1:3]:")
print(newm[1:2,1:3])

print("\n10. new_array[0:3,0:0]:")
print(new_array[0:3,0:0])

print("\n11. new_array[0:2,0:1]:")
print(new_array[0:2,0:1])

print("\n12. new_array[0:3,0:1]:")
print(new_array[0:3,0:1])

print("\n13. new_array[1:3]:")
print(new_array[1:3])
```
