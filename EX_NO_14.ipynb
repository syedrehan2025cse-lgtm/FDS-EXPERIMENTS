```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler, MinMaxScaler

# Experiment No. 14
# Understanding Feature Scaling

# 1. Create the dataset
data = {
    'Country': ['France', 'Spain', 'Germany', 'Spain', 'Germany',
                'France', 'Spain', 'France', np.nan, 'France'],
    'Age': [44, 27, 30, 38, 40, 35, np.nan, 48, 50, 37],
    'Salary': [72000, 48000, 54000, 61000, np.nan,
               58000, 52000, 79000, 83000, 67000],
    'Purchased': ['No', 'Yes', 'No', 'No', 'Yes',
                  'Yes', 'No', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)

print("ORIGINAL DATASET")
print(df)

# 2. Separate features and label
features = df.iloc[:, :-1].copy()
label = df.iloc[:, -1].copy()

# 3. Handle missing categorical values
features['Country'] = features['Country'].fillna(
    features['Country'].mode()[0]
)

# 4. Handle missing numerical values using mean
imputer = SimpleImputer(strategy='mean')
features[['Age', 'Salary']] = imputer.fit_transform(
    features[['Age', 'Salary']]
)

print("\nDATASET AFTER HANDLING MISSING VALUES")
print(features)

# 5. Encode the Country column
encoder = OneHotEncoder(sparse_output=False)
country_encoded = encoder.fit_transform(features[['Country']])

country_columns = encoder.get_feature_names_out(['Country'])
country_df = pd.DataFrame(
    country_encoded,
    columns=country_columns
)

# 6. Combine encoded country with numerical features
final_set = pd.concat(
    [
        country_df.reset_index(drop=True),
        features[['Age', 'Salary']].reset_index(drop=True)
    ],
    axis=1
)

print("\nDATASET AFTER ONE-HOT ENCODING")
print(final_set)

# 7. Apply Standard Scaling
scaler = StandardScaler()
feat_standard_scaler = scaler.fit_transform(final_set)

standard_df = pd.DataFrame(
    feat_standard_scaler,
    columns=final_set.columns
)

print("\nFEATURES AFTER STANDARD SCALING")
print(standard_df.round(3))

# 8. Apply Min-Max Scaling
minmax = MinMaxScaler(feature_range=(0, 1))
feat_minmax_scaler = minmax.fit_transform(final_set)

minmax_df = pd.DataFrame(
    feat_minmax_scaler,
    columns=final_set.columns
)

print("\nFEATURES AFTER MIN-MAX SCALING")
print(minmax_df.round(3))

print("\nEXPERIMENT NO. 14 COMPLETED SUCCESSFULLY")





```
